# Config
This script is used to get config information such as the canvas positions, palette colors etc.

```python
from robot import SerialPrinter, RobotCalibrator
from datatypes import RobotCalibration
```

```python
printer = SerialPrinter()
printer.connect()
```

```python
robot_calibrator = RobotCalibrator(printer)
```

## Canvas positions

```python
x_left, y_bottom, z_bottom = robot_calibrator.input_position(query="Canvas Bottom Left", color=False)[0]
x_right, y_top, _ = robot_calibrator.input_position(query="Canvas Top Right", color=False, ask_z=False)[0]
x_water, y_water, z_water = robot_calibrator.input_position(query="Water Reservoir", color=False, ask_z=True)[0]
_, _, safe_height = robot_calibrator.input_position(query="Safe z height", color=False, ask_z=True)[0]


my_robot_calibration = RobotCalibration(
    (x_left, y_top, z_bottom),
    (x_right, y_top, z_bottom),
    (x_left, y_bottom, z_bottom),
    (x_right, y_bottom, z_bottom),
    water_reservoir=(x_water, y_water, z_water),
    safe_height=safe_height,
)
```

## Set Color Palette

```python
color_count = 2

for i in range(color_count):
    position, color = robot_calibrator.input_position(query=f"Color {i+1}", color=True)
    my_robot_calibration.color_palette.add_color(position, color)
```

## Export

```python
my_robot_calibration.dump("data/my_robot_calibration.json")
```
