```python
import cv2
import matplotlib.pyplot as plt

# Load image via OpenCV
img = cv2.imread("C:/Users/Simon/Documents/projecten/paint-bot/images/elephant.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Initialize OpenCV's Fine Grained Saliency
saliency = cv2.saliency.StaticSaliencyFineGrained_create()
success, saliency_map = saliency.computeSaliency(img_rgb)

# saliency_map is already normalized between 0.0 and 1.0
plt.imshow(saliency_map, cmap='magma')
plt.show()
```
