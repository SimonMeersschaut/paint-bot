```python
from robot import SerialPrinter, Printer
from robot import load_brush, execute_stroke, scrub, water_brush
from robot import Camera
from datatypes import RobotCalibration, StrokeSequence, StrokePath
```

```python
printer = SerialPrinter()
printer.connect()
```

```python
Camera.start()
```

```python
FEED_RATE_TRAVEL = 3000

FEED_RATE_WET = 1500
FEED_RATE_LOAD = 700
FEED_RATE_PAINT = 1200
```

## Load Data

```python
my_calibration = RobotCalibration.load("data/my_robot_calibration.json")
color_index = 2
```

## Load Brush

```python
water_brush(printer, my_calibration)
load_brush(printer, my_calibration, color_index)
```

## Execute Stroke

```python
my_stroke_sequence = StrokeSequence((235, 130))

START_X, START_Y = (50, 10 + color_index*40)

for i in range(0, 7):
    x_begin = START_X + i *10
    y_begin = START_Y
    y_end = START_Y + 20

    my_stroke_sequence.strokes.append(
        StrokePath(
            None,
            [(x_begin, y_begin), (x_begin, y_end)],
            pigment=1
        )
    )


for i in range(len(my_stroke_sequence.strokes)):
    execute_stroke(
        printer=printer,
        robot_calibration=my_calibration,
        stroke_sequence=my_stroke_sequence,
        index=i
    )
```

```python
printer.move_to(z=my_calibration.safe_height)
print("Waiting")
printer.move_and_wait(x=0, y=150, z=my_calibration.safe_height)
print("taking picture")
```

```python
Camera.take_picture(filename=f"color_tests/color_{color_index}.png")

Camera.close()
```
