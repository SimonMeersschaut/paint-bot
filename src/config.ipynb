# Config
This script is used to get config information such as the canvas positions, palette colors etc.

```python
from robot import SerialPrinter, input_position
from datatypes import RobotCalibration
```

```python
printer = SerialPrinter()
printer.connect()
```

## Canvas positions

```python
x_left, y_bottom, z_bottom = input_position(printer, query="Canvas Bottom Left", color=False)[0]
x_right, y_top, _ = input_position(printer, query="Canvas Top Right", color=False)[0]


my_robot_calibration = RobotCalibration(
    (x_left, y_top, z_bottom),
    (x_right, y_top, z_bottom),
    (x_left, y_bottom, z_bottom),
    (x_right, y_bottom, z_bottom),
)
```

## Export

```python
my_robot_calibration.dump("my_robot_calibration.json")
```

## Set Color Palette

```python
color_count = 2

for i in range(color_count):
    position, color = input_position(printer, query=f"Color {i+1}", color=True)
    my_robot_calibration.color_palette.add_color(position, color)
```

```python
my_robot_calibration = RobotCalibration.load("my_robot_calibration.json")
```

## Water Reservoir

```python
position, _ = input_position(printer, query=f"Water Reservoir", color=False)
my_robot_calibration.water_reservoir = position
```
